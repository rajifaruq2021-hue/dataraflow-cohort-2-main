- Create a base class `Shape` with a method `area()` that raises `NotImplementedError`.
- Create subclasses `Rectangle` and `Circle` that implement `area()` appropriately.
- Instantiate both and print their areas.

````python

# Answer: 
import math


# Base (Parent) Class
class Shape:

  def area(self):
    # This acts as a safety rule reminding subclasses to define their own area logic
    raise NotImplementedError("Subclass must implement abstract method")


# Subclass Rectangle inheriting from Shape
class Rectangle(Shape):

  def __init__(self, width, height):
    self.width = width
    self.height = height

  def area(self):
    # Rectangle area formula: width * height
    return self.width * self.height


# Subclass Circle inheriting from Shape
class Circle(Shape):

  def __init__(self, radius):
    self.radius = radius

  def area(self):
    # Circle area formula: pi * radius squared
    return math.pi * (self.radius**2)


# Instantiate and test both shapes!
rect = Rectangle(4, 5)
circ = Circle(3)

print("Rectangle Area:", rect.area())  # Output: 20
print("Circle Area:", round(circ.area(), 2))  # Output: ~28.27


# Explanation:
# import math: we import Python's built-in math toolkit so we can use the exact value of Pi (math.pi) for our circle calculations.
# raise NotImplementedError: This is like a friendly warning sign from the parent class saying, "If you are a shape, you must write your own math formula for area()"
# Method Overriding: Even though both classes inherit from Shape, each child implements its own version of area() tailored specifically to its unique geometry
````

- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python

# Answer: 
# Base (Parent) Class: BankAccount
class BankAccount:

  def __init__(self, owner, balance=0.0):
    self.owner = owner
    self.balance = balance

  def deposit(self, amount):
    if amount > 0:
      self.balance += amount
      print(
          f"[{self.owner}] Deposited ${amount:.2f}. New Balance:"
          f" ${self.balance:.2f}"
      )
    else:
      print("Deposit amount must be greater than zero!")

  def withdraw(self, amount):
    if 0 < amount <= self.balance:
      self.balance -= amount
      print(
          f"[{self.owner}] Withdrew ${amount:.2f}. New Balance:"
          f" ${self.balance:.2f}"
      )
    else:
      print(f"[{self.owner}] Withdrawal failed: Insufficient funds or invalid amount.")


# Subclass (Child) Class: SavingsAccount inheriting from BankAccount
class SavingsAccount(BankAccount):

  def __init__(self, owner, balance=0.0, interest_rate=0.05):
    # super() reaches up to the parent BankAccount to initialize owner and balance
    super().__init__(owner, balance)
    self.interest_rate = interest_rate  # Adding a brand-new unique property!

  def add_interest(self):
    interest_earned = self.balance * self.interest_rate
    self.balance += interest_earned
    print(
        f"[{self.owner}] Interest added! Earned ${interest_earned:.2f}. New"
        f" Balance: ${self.balance:.2f}"
    )


# --- Demonstrating Our Banking System ---
# 1. Create a Savings Account for our friend Alice with an initial balance and interest rate
my_savings = SavingsAccount("Alice", balance=1000.0, interest_rate=0.04)

# 2. Deposit some extra cash
my_savings.deposit(500.0)

# 3. Apply interest calculation
my_savings.add_interest()

# 4. Withdraw some money
my_savings.withdraw(200.0)


# Explanation:
# super().__init__(owner, balance): Instead of rewriting code to set up the owner and balance, we let the parent BankAccount handle the heavy lifting while we focus on our new features.
# self.interest_rate = interest_rate: A custom attribute unique only to the SavingsAccount child class. 
# Safety Checks (if 0 < amount <= self.balance): Ensures our program doesn't allow a negative withdrawal or overdrawing more money than is actually in the account
````

- Create a base class `Vehicle` with attributes `brand` and `model`, and a method `display_info()`.

- Create another subclass `ElectricCar` (child of `Car`) that adds `battery_capacity` and overrides `display_info()` to include all details.

````python

# Answer: 
# Grandparent Class: Vehicle
class Vehicle:

  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def display_info(self):
    print(f"Vehicle Brand: {self.brand}, Model: {self.model}")


# Parent Class: Car (inherits from Vehicle)
class Car(Vehicle):

  def __init__(self, brand, model, doors):
    super().__init__(brand, model)  # Inherits brand and model from Vehicle
    self.doors = doors  # Adds a new property specific to cars

  def display_info(self):
    # Overriding display_info to include door count
    print(f"Car Brand: {self.brand}, Model: {self.model}, Doors: {self.doors}")


# Child Class: ElectricCar (inherits from Car, which inherits from Vehicle)
class ElectricCar(Car):

  def __init__(self, brand, model, doors, battery_capacity):
    super().__init__(brand, model, doors)  # Reaches up through Car to Vehicle
    self.battery_capacity = battery_capacity  # Adds battery specification

  def display_info(self):
    # Overriding display_info to include everything, including battery power!
    print(
        f"Electric Car: {self.brand} {self.model} | Doors: {self.doors} |"
        f" Battery: {self.battery_capacity}kWh"
    )


# --- Testing Our Multi-Level Electric Car ---
# Create an ElectricCar object[cite: 6]
my_ev = ElectricCar("Tesla", "Model S", 4, 100)

# Call display_info() to see the entire family tree of attributes in action![cite: 6]
my_ev.display_info()


# Explanation:
# Multi-Level Chain (ElectricCar -> Car -> Vehicle): The super() function creates a seamless elevator ride through the classes, making sure every parent's initialization code runs correctly.  
# Method Overriding: Even though Vehicle, Car, and ElectricCar all have a method named display_info(), Python automatically executes the most specific version belonging to the ElectricCar object
````

- Take a list of numbers `[10, 20, 30, 40, 50]`.  
- Convert it into an iterator using `iter()` and retrieve each element using `next()`.  
- Handle the `StopIteration` exception gracefully when there are no more items.

````python
# Answer: 
# Our list of numbers
numbers = [10, 20, 30, 40, 50]

# Convert the list into an iterator using iter()
my_iter = iter(numbers)

# Retrieve items one by one safely using a loop and try-except
while True:
  try:
    print(next(my_iter))
  except StopIteration:
    print("Reached the end of the iterator safely!")
    break

# Explanation:
# iter(numbers): This converts our list into an iterator object, allowing us to traverse it one item at a time.
# next(my_iter): This fetches the next item from the iterator. If there are no more items, it raises a StopIteration exception.
# try-except block: This is a safety net that catches the StopIteration exception, preventing our program from crashing and allowing us to handle the end of the iteration gracefully. 
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
# Answer: 
class CountDown:

  def __init__(self, n):
    self.n = n

  def __iter__(self):
    # Returns the iterator object itself
    return self

  def __next__(self):
    if self.n > 0:
      current_value = self.n
      self.n -= 1
      return current_value
    else:
      # Stops the iteration when we hit zero
      raise StopIteration


# Testing our CountDown iterator in a for loop!
print("Countdown starting:")
for num in CountDown(5):
  print(num)

# Explanation:
# __iter__(self): This method returns the iterator object itself, allowing it to be used in a for loop or any other iteration context.
# __next__(self): This method defines how to get the next item in the sequence.     
````

- Create a class `EvenNumbers` that generates even numbers starting from 0.  
- The iterator should be infinite (no `StopIteration`).  
- Demonstrate usage by printing the first 10 even numbers using a loop and `break`.

````python
# Answer: 
class EvenNumbers:

  def __init__(self):
    self.num = 0

  def __iter__(self):
    return self

  def __next__(self):
    current_value = self.num
    self.num += 2  # Jump to the next even number
    return current_value


# Demonstrate usage by printing the first 10 even numbers
evens = EvenNumbers()
my_iter = iter(evens)

print("First 10 Even Numbers:")
count = 0
for num in my_iter:
  print(num)
  count += 1
  if count >= 10:  # Break out of the infinite loop once we hit 10!
    break

  
# Explanation / Comments:
# __next__(self): This method generates the next even number by adding 2 to the current value each time it's called.        
# This is cool because the class doesn't have a built-in stopping condition, it can theoretically run forever
# We use an external counter and a break statement to gracefully stop whenever we choose.
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
# Answer: 
class FileLineReader:

  def __init__(self, filename):
    self.filename = filename
    self.file = None

  def __iter__(self):
    # Open the file in read mode when iteration starts
    self.file = open(self.filename, "r")
    return self

  def __next__(self):
    # Read the next line from the file
    line = self.file.readline()

    # If the line is empty, we have reached the end of the file!
    if not line:
      self.file.close()  # Always clean up and close your files
      raise StopIteration

    # Return the line stripped of trailing newline characters
    return line.strip()


# --- How to test it ---
# (First, we need to make sure we have a file named 'sample.txt' in our folder!)
# reader = FileLineReader("sample.txt")
# for line in reader:
#     print(line)


# Explanation:
# The FileLineReader class is an iterator that reads a file line by line.
# __iter__(self): Opens the file and returns the iterator object.  
# __next__(self): Reads the next line from the file. If there are no more lines, it raises StopIteration to signal the end of iteration.    
# This allows us to process massive files line-by-line without loading the entire document into our computer's memory all at once         
````

- Create a global variable `total = 0`.

````python

# Answer: 
## Initialize our global total tracker at the module level
total = 0


def add_to_total(n):
  # Declare that we want to modify the global 'total' variable directly
  global total

  # Add 'n' to our running global total accumulator
  total += n
  print(f"Added {n}. Current global total value:", total)


# --- Execution ---
add_to_total(5)
add_to_total(15)
add_to_total(10)
print("Final global total after all calls:", total)

# Explanation:
#This demonstrates how we persist state across multiple function calls by leveraging the global keyword. Ordinarily, variables created inside functions are temporary. By declaring global total, we instruct Python to link our function directly to the module-level variable, allowing us to accumulate and preserve values continuously across our program.
#
````

- Create a function `outer()` with a variable `message = "Hi"`.  
- Inside `outer()`, create a nested function `inner()` that modifies `message` using the `nonlocal` keyword.  
- Call `inner()` from inside `outer()` and print the updated `message`.

````python

# Answer: 
def outer():
  # Define a variable in the enclosing outer function scope
  message = "Hi"

  def inner():
    # Declare that we are modifying the nonlocal variable from the enclosing scope
    nonlocal message

    # Update the enclosing variable's value
    message = "Hello from inner scope!"
    print("Inside inner function, modified message:", message)

  print("Before inner execution, message is:", message)
  inner()
  print("After inner execution, message is:", message)


# --- Execution ---
outer()

# Explanation:
# This specifies that when we work with nested functions, the nonlocal keyword empowers an inner function to modify variables residing in its parent enclosing function. 
# Without nonlocal, Python would treat message inside inner() as a separate local variable, leaving the outer variable untouched.
# This allows us to share and mutate state vertically through function layer
````

  - Then creates a local variable named `len` with a value of `100`.  
  - Tries to call `len()` again and observes the error.  
- Explain what happened and how it relates to built-in scope.

````python
# Answer: 
def show_len_behavior():
  sample_list = [10, 20, 30, 40]

  # 1. Successfully call the built-in len() function on our list
  print("Initial length check using built-in len():", len(sample_list))

  # 2. Create a local variable named 'len' and assign it an integer value
  len = 100
  print("Assigned local variable len:", len)

  # 3. Trying to call len() again raises a TypeError because 'len' is now an integer, not a function!
  try:
    print(len(sample_list))
  except TypeError as e:
    print("Caught expected TypeError due to built-in shadowing:", e)


# --- Execution ---
show_len_behavior()

# Explanation:
# This illustrates how Python resolves names using the LEGB (Local, Enclosing, Global, Built-in) rule. 
# When we assign a value to a name like len inside our function, Python creates a local variable that completely shadows Python's built-in len() function within that scope. 
# Once shadowed, Python looks up len locally, finds the integer 100, and raises a TypeError when we attempt to invoke it as a function
````

Create an iterator class `EvenNumbers` that takes a limit and iterates over even numbers only. Use it in a `for` loop.

````python
# Answer: 
class EvenNumbers:

  def __init__(self, limit):
    self.limit = limit
    self.current = 0  # Starting point for our even numbers

  def __iter__(self):
    # Return our iterator object itself to initialize the looping protocol
    return self

  def __next__(self):
    # Check if our current number is within our specified boundary limit
    if self.current <= self.limit:
      num = self.current
      self.current += 2  # Increment by 2 to stay strictly on even numbers
      return num
    else:
      # Stop iteration gracefully once we cross our limit
      raise StopIteration


# --- Execution ---
print("Iterating over even numbers up to 10:")
for even_num in EvenNumbers(10):
  print(even_num)


# Explanation:
# This illustrates how we construct a custom iterator using the iterator protocol methods __iter__() and __next__(). 
# By tracking our position with self.current up to our self.limit, our class yields sequential even numbers until completion, triggering StopIteration so our for loop terminates cleanly.
````

Create two classes: `Flyer` (method `move()` returns `"Flying"`) and `Swimmer` (method `move()` returns `"Swimming"`).  
Create a class `Duck` inheriting from both.  
- Use `super()` to decide which `move()` executes.  
- Explain the **Method Resolution Order (MRO)**.  

````python
# Answer: 
class Flyer:

  def move(self):
    return "Flying"


class Swimmer:

  def move(self):
    return "Swimming"


class Duck(Flyer, Swimmer):

  def move(self):
    # Use super() to delegate movement according to Python's MRO chain
    return f"Duck is {super().move()}"


# --- Execution ---
my_duck = Duck()
print(my_duck.move())
print("Method Resolution Order (MRO) path:", Duck.__mro__)


# Explanation:
# This explains how multiple inheritance resolves method lookups using Python's Method Resolution Order (MRO). 
# When a class inherits from multiple parents, Python follows a strict linearized path determined by the C3 linearization algorithm. 
# Because Flyer is listed first in Duck(Flyer, Swimmer), calling super().move() delegates directly to Flyer.move()
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
# Answer: 
class Fibonacci:

  def __init__(self, limit):
    self.limit = limit
    self.count = 0
    self.a = 0
    self.b = 1

  def __iter__(self):
    return self

  def __next__(self):
    if self.count < self.limit:
      if self.count == 0:
        current = self.a
      elif self.count == 1:
        current = self.b
      else:
        current = self.a + self.b
        self.a = self.b
        self.b = current
      self.count += 1
      return current
    else:
      raise StopIteration


# --- Execution ---
print("First 7 Fibonacci numbers generated via iterator:")
for fib_num in Fibonacci(7):
  print(fib_num)


# Explanation:
# This demonstrates how we implement a mathematical state-machine iterator. 
# By maintaining previous sequence terms (self.a and self.b) and a tracking count up to our limit, our class computes and yields Fibonacci numbers one by one without wasting system memory.
````

# Library Management System  
  
## Scenario  
You are building a small library system to manage books, members, and borrowing operations. The system should support multiple types of books, track borrowed books, and allow iteration over members and books. The system should also demonstrate proper use of scopes and encapsulation.  
  
## Requirements and Concepts to Cover  
  
### 1. OOP & Inheritance  
- Create a base class `Book` with attributes: `title`, `author`, `year`.  
- Create subclasses:  
  - `EBook` with `file_size` and `format`.  
  - `PrintedBook` with `weight` and `dimensions`.  
- Create a base class `Member` with `name` and `member_id`.  
- Create subclasses:  
  - `StudentMember` with `grade_level`.  
  - `TeacherMember` with `department`.  
  
### 2. Polymorphism  
- Define a method `get_description()` in `Book` and override it in both `EBook` and `PrintedBook`.  
- Define a method `borrow_book()` in `Member` that behaves differently for `StudentMember` and `TeacherMember` (e.g., different borrowing limits).  
  
### 3. Iterators  
- Implement an iterator for the `Library` class to iterate over all books.  
- Implement an iterator for the `Member` class to iterate over borrowed books.  
  
### 4. Scope  
- Use private and protected attributes to demonstrate scope:  
  - `_borrowed_books` in `Member` (protected).  
  - `__library_name` in `Library` (private).  
- Provide getter and setter methods where appropriate to access private/protected attributes.  
  
### 5. Extra Features  
- Track which books are available and which are borrowed.  
- Allow searching books by title or author.  
- Count the total number of books borrowed by all members (use a class-level variable).  

````python
# Answer: 
# ---------------------------------------------------------
# 1. OOP & INHERITANCE: BOOKS HIERARCHY
# ---------------------------------------------------------


class Book:
  # Base class representing universal book attributes
  def __init__(self, title, author, year):
    self.title = title
    self.author = author
    self.year = year
    self.is_borrowed = False  # Track availability state

  def get_description(self):
    return f"'{self.title}' by {self.author}, published in {self.year}"


class EBook(Book):
  # Subclass for electronic books adding file attributes
  def __init__(self, title, author, year, file_size, format_type):
    super().__init__(title, author, year)
    self.file_size = file_size
    self.format_type = format_type

  def get_description(self):
    # Polymorphic override providing digital details
    return (
        f"[EBook] '{self.title}' by {self.author} ({self.year}) |"
        f" {self.file_size}MB, Format: {self.format_type}"
    )


class PrintedBook(Book):
  # Subclass for physical books adding weight and dimensions
  def __init__(self, title, author, year, weight, dimensions):
    super().__init__(title, author, year)
    self.weight = weight
    self.dimensions = dimensions

  def get_description(self):
    # Polymorphic override providing physical details
    return (
        f"[PrintedBook] '{self.title}' by {self.author} ({self.year}) |"
        f" Weight: {self.weight}g, Dims: {self.dimensions}"
    )


# ---------------------------------------------------------
# 1 & 4. MEMBERS HIERARCHY & SCOPE ENCAPSULATION
# ---------------------------------------------------------


class Member:
  # Class-level variable to track total books borrowed across all members
  total_books_borrowed_globally = 0

  def __init__(self, name, member_id, borrowing_limit):
    self.name = name
    self.member_id = member_id
    self.borrowing_limit = borrowing_limit
    self._borrowed_books = []  # Protected attribute tracking borrowed books

  def borrow_book(self, book):
    # Base borrowing logic overridden by specialized member types
    if book.is_borrowed:
      print(
          f"Sorry {self.name}, '{book.title}' is already currently borrowed!"
      )
      return False

    if len(self._borrowed_books) >= self.borrowing_limit:
      print(
          f"Borrow limit reached for {self.name}! Limit is"
          f" {self.borrowing_limit}."
      )
      return False

    book.is_borrowed = True
    self._borrowed_books.append(book)
    Member.total_books_borrowed_globally += 1
    print(f"{self.name} successfully borrowed '{book.title}'.")
    return True

  def return_book(self, book):
    if book in self._borrowed_books:
      book.is_borrowed = False
      self._borrowed_books.remove(book)
      Member.total_books_borrowed_globally -= 1
      print(f"{self.name} returned '{book.title}'.")
    else:
      print(f"{self.name} does not have '{book.title}' checked out.")

  # Getter method for protected attribute access (Encapsulation)
  def get_borrowed_books(self):
    return self._borrowed_books

  def __iter__(self):
    # 3. ITERATORS: Implement an iterator for Member to iterate over borrowed books
    return iter(self._borrowed_books)


class StudentMember(Member):
  def __init__(self, name, member_id, grade_level):
    super().__init__(name, member_id, borrowing_limit=3)  # Student limit
    self.grade_level = grade_level

  def borrow_book(self, book):
    # 2. POLYMORPHISM: Specialized borrowing rule for students
    print(f"[Student Privilege Check for {self.name}]")
    return super().borrow_book(book)


class TeacherMember(Member):
  def __init__(self, name, member_id, department):
    super().__init__(name, member_id, borrowing_limit=10)  # Higher teacher limit
    self.department = department

  def borrow_book(self, book):
    # 2. POLYMORPHISM: Specialized borrowing rule for teachers
    print(f"[Teacher Privilege Check for {self.name} - Dept: {self.department}]")
    return super().borrow_book(book)


# ---------------------------------------------------------
# 3 & 4. LIBRARY CLASS WITH PRIVATE SCOPE & ITERATOR PROTOCOL
# ---------------------------------------------------------


class Library:

  def __init__(self, name):
    self.library_name = name
    self._books = []
    self.__library_pin = 9876  # Private attribute demonstrating strict scope encapsulation

  # Getter for private library PIN securely
  def get_library_pin(self):
    return self.__library_pin

  # Setter for private library PIN securely
  def set_library_pin(self, new_pin):
    if isinstance(new_pin, int) and len(str(new_pin)) == 4:
      self.__library_pin = new_pin
      print("Library security PIN updated successfully.")
    else:
      print("Invalid PIN format. Must be a 4-digit integer.")

  def add_book(self, book):
    self._books.append(book)

  def search_books(self, query):
    # 5. EXTRA FEATURES: Search books by title or author
    results = []
    query_lower = query.lower()
    for book in self._books:
      if (
          query_lower in book.title.lower()
          or query_lower in book.author.lower()
      ):
        results.append(book)
    return results

  def __iter__(self):
    # 3. ITERATORS: Implement an iterator for the Library class to iterate over all books
    return LibraryIterator(self._books)


# Helper iterator class for Library collection traversal
class LibraryIterator:

  def __init__(self, books):
    self._books = books
    self._index = 0

  def __iter__(self):
    return self

  def __next__(self):
    if self._index < len(self._books):
      book = self._books[self._index]
      self._index += 1
      return book
    else:
      raise StopIteration


# ---------------------------------------------------------
# DEMONSTRATING OUR COMPLETE SYSTEM WORKFLOW 
# ---------------------------------------------------------

# Initialize our library
my_library = Library("City Central Hub")

# Create books (EBooks and PrintedBooks)
b1 = EBook("Python Masterclass", "Guido van Rossum", 2024, 15.4, "PDF")
b2 = PrintedBook("Data Structures Deep Dive", "Alan Turing", 2021, 450, "20x15cm")
b3 = EBook("Artificial Intelligence 101", "Ada Lovelace", 2025, 22.1, "EPUB")

my_library.add_book(b1)
my_library.add_book(b2)
my_library.add_book(b3)

# Create members
student = StudentMember("Alice", "S101", "Sophomore")
teacher = TeacherMember("Dr. Bob", "T201", "Computer Science")

# Demonstrate searching books
print("--- Searching for 'Python' ---")
search_results = my_library.search_books("Python")
for match in search_results:
  print(match.get_description())

# Demonstrate borrowing operations (Polymorphic limits)
print("\n--- Borrowing Operations ---")
teacher.borrow_book(b1)
student.borrow_book(b1)  # Should fail because Dr. Bob already borrowed it!
student.borrow_book(b2)

# Demonstrate Iteration over Library books
print("\n--- Iterating Over All Library Books ---")
for book_item in my_library:
  print("-", book_item.get_description(), f"[Borrowed: {book_item.is_borrowed}]")

# Demonstrate Member Iteration over borrowed books
print("\n--- Iterating Over Student's Borrowed Books ---")
for borrowed in student:
  print("Student holds:", borrowed.title)

# Display global class-level variable tracking total borrowed books
print(
    "\nTotal books currently borrowed globally across all members:"
    f" {Member.total_books_borrowed_globally}"
)



# Explanation:

# Explanation for OOP & Inheritance:
# This demonstrates how we establish a clean parent-child relationship where Book acts as the universal template 
# while EBook and PrintedBook inherit shared attributes (title, author, year) while introducing specialized digital or physical properties.  

# Explanation for Polymorphism:
# This illustrates how polymorphism allows both EBook and PrintedBook to implement their own customized versions of get_description(), 
# and how StudentMember and TeacherMember override borrow_book() to enforce distinct borrowing limits and privilege checks seamlessly.  

# Explanation for Iterators:
# This specifies how we build custom iteration protocols using __iter__() and __next__() for both our Library class (to traverse all books)
# and our Member class (to inspect currently checked-out materials) without exposing internal data structures directly.  

# Explanation for Scope & Encapsulation:
# This highlights how we secure sensitive data by utilizing protected attributes (_borrowed_books) 
# and private attributes (__library_pin), providing controlled access via getter and setter methods while managing global states through class-level variables like total_books_borrowed_globally.  
#
````
